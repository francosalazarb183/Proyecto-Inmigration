# 🚢 Flujo Histórico de Inmigración Italiana a EEUU. (1855 - 1900)

Aplicación interactiva diseñada para explorar y conocer el flujo de inmigrantes italianos a EEUU, respondiendo a interrogantes demográficos, temporales y logísticos de la época.

## ❓ Pregunta que responde
¿Cómo evolucionó el perfil demográfico, el volumen migratorio y las rutas de salida de los inmigrantes italianos hacia América entre 1855 y 1900?

## 📊 Los Datos
- **Origen:** Registros históricos oficiales de pasajeros italianos (1855-1900) proporcionados por la cátedra.
- **Volumen:** Cuenta con 845368 registros de inmigrantes italianos y cuenta con 15 columnas.
- **Estructura de cada fila:** Cada registro detalla la información individual del pasajero ('lista', 'apellido', 'nombre', 'edad', 'sexo', 'oficio', 'sabe_leer','pais', 'ultima_residencia', 'destino', 'tipo_pasaje', 'barco','puerto_salida', 'fecha', 'año').

## 🛠️ ¿Qué se limpió y por qué?
- Corrección de puertos: Se corrigio un código del puerto "089" para evitar nulos en las rutas marítimas.
- Tratamiento avanzado de la edad: Se integró el archivo de códigos de edad (`AGECOD`), interpretando correctamente los códigos especiales (como los de la serie 900 que indican meses de vida en bebés) y evitando eliminar registros válidos mediante filtros erróneos de mayores de 110 años.
- Tipos de datos y nulos: Se revisaron y se reemplazaron por "DESCONOCIDO" todos los indicios de nulos/palabras equivalentes a nulos, como "UKNKNOWN","-", "800","U","N". Tambien se reemplazaron minúsculas por mayúsculas.

## 🔍 Tres Hallazgos Clave
1. Pico migratorio de fin de siglo: El volumen de pasajeros experimenta un crecimiento exponencial abrupto a partir de la década de 1890, reflejando la crisis socioeconómica europea y la consolidación de las rutas transatlánticas.
2. Predominio demográfico activo: Se observa una fuerte concentración de pasajeros jóvenes adultos, con una marcada proporción de hombres en búsqueda de oportunidades laborales.
3. Concentración de puertos de salida: Los grandes nodos portuarios del norte y sur de Italia (como Génova y Nápoles) concentraron la inmensa mayoría de las partidas, evidenciando los corredores logísticos principales de la época.
